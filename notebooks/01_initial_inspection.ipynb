```python
import pandas as pd

# adjust path if your Excel has another name/extension
path = "../data/raw/Online Retail.xlsx"  # or .xls/.csv
df = pd.read_excel(path, engine="openpyxl")  # or pd.read_csv if CSV

# quick shape & head
print("rows, cols:", df.shape)
display(df.head())

# basic types & non-null counts
display(df.info())

# convert date
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')

# add revenue
df['Revenue'] = df['Quantity'] * df['UnitPrice']

# quick checks
print("Negative quantity rows:", (df['Quantity'] < 0).sum())
print("Zero or negative unit price rows:", (df['UnitPrice'] <= 0).sum())
print("Missing CustomerID rows:", df['CustomerID'].isna().sum(),
      f"({df['CustomerID'].isna().mean():.2%})")

# invoices that look like cancellations
c_inv_mask = df['InvoiceNo'].astype(str).str.startswith('C', na=False)
print("Invoices starting with 'C' (likely cancellations):", c_inv_mask.sum())

# top products by frequency and revenue
print("\nTop 10 SKUs by count:")
display(df['StockCode'].value_counts().head(10))
print("\nTop 10 SKUs by revenue:")
display(df.groupby('StockCode')['Revenue'].sum().sort_values(ascending=False).head(10))
```
